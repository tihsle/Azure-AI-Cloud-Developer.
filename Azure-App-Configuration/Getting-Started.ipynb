### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()
app_config_conn_string = os.getenv("AZURE_APPCONFIG_CONNECTION_STRING")
```

### Creating the App Configuration Client

```python
from azure.appconfiguration.provider import load

# creating the Azure App Configuration loader object
config = load(
    connection_string=app_config_conn_string
)
```

### Access All the values using dictionary syntax

```python
azure_openai_endpoint = config["AzureOpenAI:Endpoint"]
azure_openai_api_key = config["AzureOpenAI:Key"]
azure_openai_model_name = config["AzureOpenAI:ModelName"]

print(f"Azure OpenAI Endpoint: {azure_openai_endpoint}")
print(f"Azure OpenAI API Key: {azure_openai_api_key}")
print(f"Azure OpenAI Model Name: {azure_openai_model_name}")
```

### Select and filter keys with SettingSelector

```python
from azure.appconfiguration.provider import SettingSelector

# creating the Selector Filter object
selects = [
    SettingSelector(key_filter="AzureOpenAI:*", label_filter="Production")
]

# recreating the app config provider with the selector filter object as a parameter
config = load(
    connection_string=app_config_conn_string,
    selects=selects
)

# fetching and printing the production configuration values
print(f"Azure OpenAI Endpoint (Production): {config["AzureOpenAI:Endpoint"]}")
print(f"Azure OpenAI API Key (Production): {config["AzureOpenAI:Key"]}")
print(f"Azure OpenAI Model Name (Production): {config["AzureOpenAI:ModelName"]}")
```
