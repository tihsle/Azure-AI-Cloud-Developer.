### Setting up the Environment

```python
import os 
from dotenv import load_dotenv

load_dotenv()

endpoint = os.getenv("COSMOSDB_ENDPOINT")
key = os.getenv("COSMOSDB_KEY")
database_name = os.getenv("DATABASE_NAME")
container_name = os.getenv("CONTAINER_NAME")
```

### Creating the CosmosDB Client

```python
from azure.cosmos import CosmosClient

client = CosmosClient(endpoint, key)
```

### Navigate the Resource Hierarchy

```python
database = client.get_database_client(database_name)
container = database.get_container_client(container_name)
```

### Uploading Data to Azure CosmosDB via SDK

```python
from azure.cosmos import exceptions
import json

with open("./Data/food_dataset.json", "r") as f:
    data = json.load(f)

try:
    for item in data:
        response = container.create_item(body=item) # use upsert_item to update existing items or create new ones
        print(f"Inserted item with id: {response['id']}")
except exceptions.CosmosResourceExistsError:
    print("Item already exists in the container.")
```

### Read Items Efficiently

```python
item = container.read_item(
    item="food0001",
    partition_key="Smoothies"
)

print("Item read from CosmosDB:")
print(json.dumps(item, indent=4))
```
