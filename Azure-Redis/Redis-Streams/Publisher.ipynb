### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()
redis_hostname = os.getenv("REDIS_HOSTNAME")
redis_password = os.getenv("REDIS_PASSWORD")
```

### Setting up the Redis Client

```python
import redis

r = redis.Redis(
    host = redis_hostname,
    port = 10000,
    ssl = True,
    decode_responses = True,
    password = redis_password
)
```

### Publishing Message to a Queue

```python
task_id = r.xadd('ai:inference:queue', {
    'prompt': "Tell me something about Redis Task Streams in a Pub/Sub Model",
    'user_id': "user-1001",
})

print("task created with the id: {}".format(task_id))
```
