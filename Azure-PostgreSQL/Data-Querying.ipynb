### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()
host = os.getenv("DATABASE_HOSTNAME")
db_name = os.getenv("DATABASE_NAME")
username = os.getenv("DATABASE_USERNAME")
password = os.getenv("DATABASE_PASSWORD")
```

### Create a Connection Pool

```python
from psycopg_pool import ConnectionPool

pool = ConnectionPool(
    conninfo=(
        f"host={host} "
        f"dbname={db_name} "
        f"user={username} "
        f"password={password} "
        f"sslmode=require"
    ),
    min_size=2,
    max_size=10
)

pool.wait()

print("Connection pool created successfully")
```

### Use the Connection Pool to Execute a SELECT Query

```python
with pool.connection() as conn:
    with conn.cursor() as cur:
        cur.execute("""
            SELECT *
            FROM ESG.Companies
        """)

        results = cur.fetchall()

        for row in results:
            print(row)
```

### Filter Data Using WHERE

```python
with pool.connection() as conn:
    with conn.cursor() as cur:
        cur.execute("""
        SELECT *
        FROM ESG.EmissionRecords
        WHERE Scope = %s
        """,
        (1,))

        results = cur.fetchall()

        for row in results:
            print(row)
```

### Filter Emissions Greater than 500

```python
with pool.connection() as conn:
    with conn.cursor() as cur:
        cur.execute("""
        SELECT *
        FROM ESG.EmissionRecords
        WHERE CO2_Emissions > %s
        """,
        (500,))

        results = cur.fetchall()

        for row in results:
            print(row)
```

### Find Companies in India

```python
with pool.connection() as conn:
    with conn.cursor() as cur:
        cur.execute("""
            SELECT *
            FROM ESG.EmissionRecords
            ORDER BY CO2_Emissions DESC
            LIMIT 3
        """)

        results = cur.fetchall()

        for row in results:
            print(row)
```

### Sort by Emissions

```python
with pool.connection() as conn:
    with conn.cursor() as cur:
        cur.execute("""
            SELECT *
            FROM ESG.EmissionRecords
            ORDER BY CO2_Emissions ASC
        """)

        results = cur.fetchall()

        for row in results:
            print(row)
```

### Calculate Total Emissions per Company

```python
with pool.connection() as conn:
    with conn.cursor() as cur:
        cur.execute("""
            SELECT
                CompanyID,
                SUM(CO2_Emissions) AS TotalEmissions
            FROM ESG.EmissionRecords
            GROUP BY CompanyID
        """)

        results = cur.fetchall()

        for row in results:
            print(row)
```
